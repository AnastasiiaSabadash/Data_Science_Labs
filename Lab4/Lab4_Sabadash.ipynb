## Етап 1. Підготовка й відтворюваність

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

SEED = 42
np.random.seed(SEED)

df = sns.load_dataset("titanic")
df.head()
```

**Висновок:**  
Успішно імпортовано необхідні бібліотеки для роботи з даними та побудови графіків (`numpy`, `pandas`, `matplotlib`, `seaborn`). Зафіксовано фіксоване зерно випадкових чисел (`SEED = 42`) для забезпечення відтворюваності результатів, а також завантажено датасет Titanic та перевірено його структуру за допомогою попереднього перегляду перших рядків.

## Етап 2. Первинний аналіз даних

```python
print(df.shape)
display(df.head())
df.info()
display(df.describe(include="all"))
print("Дублікати:", df.duplicated().sum())
```

```python
missing = pd.DataFrame({
    "count": df.isna().sum(),
    "percent": df.isna().mean().mul(100),
}).sort_values("percent", ascending=False)

display(missing)
```

**Висновок:**  
Проведено первинний аналіз структури даних датасету Titanic. Досліджено типи даних, описову статистику та виявлено дублікати (107 записів). Аналіз пропущених значень показав, що найбільша кількість пропусків спостерігається у стовпчиках `deck` (~77%) та `age` (~20%).

## Етап 3. Формування ознак і цілі

```python
data = df.copy()
data["family_size"] = data["sibsp"] + data["parch"] + 1

numeric_features = ["age", "sibsp", "parch", "fare", "family_size"]
categorical_features = ["pclass", "sex", "embarked"]

X = data[numeric_features + categorical_features]
y = data["survived"]
```

Одиниця додається тому, що `sibsp` (кількість братерства/подружжя) та `parch` (кількість батьків/дітей) враховують лише *родичів* пасажира, але не саму людину. Додавання `+ 1` включає самого пасажира до загального розміру сім'ї (`family_size`).

**Висновок:**  
У результаті виконання етапу було створено нову комбіновану ознаку `family_size` та сформовано матрицю ознак `X` (із розбивкою на 5 числових і 3 категоріальні змінних) та цільовий вектор `y` (`survived`). Ознаку `pclass` залишено серед категоріальних, оскільки її значення відповідають кодам класів квитка, а не неперервній числовій шкалі.

## Етап 4. Навчальна і тестова вибірки

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED
)

print(y.mean(), y_train.mean(), y_test.mean())
```

**Висновок:**  
Завдяки використанню параметра `stratify=y` розподіл цільової змінної (частка тих, хто вижив) збережено майже однаковим у початкових даних (`y`), навчальній (`y_train`) та тестовій (`y_test`) вибірках (~38.4%). Це гарантує репрезентативність обох вибірок і запобігає дисбалансу класів під час підготовки та оцінювання моделі.

## Етап 5. Базова модель

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
baseline_pred = baseline.predict(X_test)
```

**Висновок:**  
Побудовано базову модель (`DummyClassifier`) за стратегією найчастішого класу (`most_frequent`), яка слугуватиме початковою точкою відліку (бенчмарком) для оцінки ефективності складніших моделей класифікації на тестовій вибірці.

## Етап 6. Підготовка ознак у Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features),
])

model = Pipeline([
    ("prepare", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, random_state=SEED)),
])
```

**Висновок:**  
Створено єдиний автоматизований конвеєр (`Pipeline`), який включає поетапну обробку числових ознак (заповнення медіаною та стандартизація), обробку категоріальних ознак (заповнення найчастішим значенням та кодування OneHotEncoder) і подальшу класифікацію за допомогою логістичної регресії. Такий підхід повністю виключає виток даних (data leakage) під час крос-валідації чи оцінювання.

## Етап 7. Навчання та прогноз

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]
```

**Висновок:**  
Виконано навчання моделі логістичної регресії у складі єдиного конвеєра на тренувальній вибірці `X_train` / `y_train`. Отримано прогнозовані класи для тестової вибірки (`y_pred`), а також розраховано ймовірності приналежності до позитивного класу (`y_proba`), які надалі будуть використані для оцінки якості та вибору оптимального порога класифікації.

## Етап 8. Оцінювання

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, average_precision_score,
    RocCurveDisplay, PrecisionRecallDisplay
)

def classification_metrics(y_true, y_hat):
    return {
        "accuracy": accuracy_score(y_true, y_hat),
        "precision": precision_score(y_true, y_hat, zero_division=0),
        "recall": recall_score(y_true, y_hat, zero_division=0),
        "f1": f1_score(y_true, y_hat, zero_division=0),
    }

results = pd.DataFrame([
    classification_metrics(y_test, baseline_pred),
    classification_metrics(y_test, y_pred),
], index=["DummyClassifier", "LogisticRegression"])

results.loc["LogisticRegression", "roc_auc"] = roc_auc_score(y_test, y_proba)
results.loc["LogisticRegression", "average_precision"] = average_precision_score(y_test, y_proba)
display(results.round(3))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap="Blues")
plt.title("Матриця помилок: LogisticRegression")
plt.show()

RocCurveDisplay.from_predictions(y_test, y_proba)
plt.show()

PrecisionRecallDisplay.from_predictions(y_test, y_proba)
plt.show()
```

**Висновок:**  
Проведено порівняльний аналіз якості базової моделі та логістичної регресії за допомогою метрик Accuracy, Precision, Recall, F1-score, а також ROC AUC і Average Precision (обчислених за ймовірностями). Отримана таблиця результатів та графіки (матриця помилок, ROC та PR криві) підтверджують суттєву перевагу основної моделі над базовим бенчмарком.

## Етап 9. Аналіз порога рішення

```python
threshold_rows = []
for threshold in [0.30, 0.50, 0.70]:
    pred_t = (y_proba >= threshold).astype(int)
    row = classification_metrics(y_test, pred_t)
    row["threshold"] = threshold
    threshold_rows.append(row)

threshold_results = pd.DataFrame(threshold_rows).set_index("threshold")
display(threshold_results.round(3))
```

**Аналіз компромісу та вибір порога:**
* **Як змінюється компроміс:** Зі збільшенням порога (від $0.30$ до $0.70$) кількість спрацьовувань моделі зменшується. Це призводить до зростання точності (`precision`), оскільки модель стає більш впевненою у позитивних прогнозах, але водночас падає повнота (`recall`), оскільки частина справжніх пасажирів, які вижили, пропускається.
* **Вибір порога для умовного сценарію:** За сценарієм, де пропуск пасажира, який вижив, вважається дорожчою помилкою (тобто критично важливо знайти якомога більше тих, хто вижив), слід обрати низький поріг **$0.30$**. Це забезпечить максимальний `recall` (повноту), хоча й ціною зниження загальної точності (`precision`). Зверніть увагу: це навчальний аналіз для конкретного сценарію, а не універсальний оптимальний поріг для реального продакшну.

**Висновок:**  
Проведено дослідження впливу порога класифікації на метрики якості. Наочно продемонстровано зворотний зв'язок між `precision` та `recall` і виконано обґрунтований вибір порога відповідно до умов бізнес-сценарію.

## Етап 10. Аналіз окремих помилок

```python
error_analysis = X_test.copy()
error_analysis["actual"] = y_test
error_analysis["predicted"] = y_pred
error_analysis["probability"] = y_proba

false_positive = error_analysis.query("actual == 0 and predicted == 1")
false_negative = error_analysis.query("actual == 1 and predicted == 0")

display(false_positive.sort_values("probability", ascending=False).head())
display(false_negative.sort_values("probability").head())
```

**Аналіз причин помилок та шляхи покращення:**
* **Можливі причини помилок:**
  1. *Лінійність моделі:* Логістична регресія добре працює з лінійними залежностями, але може не враховувати складні нелінійні комбінації ознак (наприклад, поєднання класу квитка, віку та статі пасажира).
  2. *Обмеженість наявних ознак:* У базовому наборі даних відсутні деякі важливі фактори (наприклад, розташування каюти на кораблі або точні обставини евакуації), які могли вплинути на порятунок.
* **Способи покращення експерименту:**
  1. *Застосування складніших моделей:* Використати алгоритми ансамблів (наприклад, Random Forest або Gradient Boosting), здатні виявляти нелінійні взаємозв'язки.
  2. *Генерація нових ознак (Feature Engineering):* Створити додаткові змінні на основі наявних (наприклад, виділити вікові категорії або витягнути додаткові дані з імен пасажирів).

**Висновок:**  
Проведено детальний аналіз окремих помилок класифікатора на тестовій вибірці. Досліджено хибно позитивні (FP) та хибно негативні (FN) прогнози з найвищим ступенем впевненості моделі, запропоновано обґрунтовані причини їх виникнення та перспективні шляхи покращення якості експерименту.

## Індивідуальне завдання
**Завдання:** Додати ознаку `is_alone` та перевірити зміну метрик.

```python
# 1. Створення нових даних з додаванням is_alone
data_inv = df.copy()
data_inv["family_size"] = data_inv["sibsp"] + data_inv["parch"] + 1
data_inv["is_alone"] = (data_inv["family_size"] == 1).astype(int)

# 2. Оновлюємо списки ознак
numeric_features_inv = ["age", "sibsp", "parch", "fare", "family_size"]
categorical_features_inv = ["pclass", "sex", "embarked", "is_alone"]

X_inv = data_inv[numeric_features_inv + categorical_features_inv]
y_inv = data_inv["survived"]

# 3. Стратифікований поділ даних
X_tr_inv, X_te_inv, y_tr_inv, y_te_inv = train_test_split(
    X_inv, y_inv, test_size=0.20, stratify=y_inv, random_state=SEED
)

# 4. Створення та навчання препроцесора і моделі
preprocessor_inv = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features_inv),
    ("cat", categorical_pipeline, categorical_features_inv),
])

model_inv = Pipeline([
    ("prepare", preprocessor_inv),
    ("classifier", LogisticRegression(max_iter=1000, random_state=SEED)),
])

model_inv.fit(X_tr_inv, y_tr_inv)

# 5. Отримання прогнозів
y_pred_inv = model_inv.predict(X_te_inv)
y_proba_inv = model_inv.predict_proba(X_te_inv)[:, 1]

# 6. Порівняння метрик до та після додавання is_alone
metrics_before = classification_metrics(y_test, y_pred)
metrics_before["roc_auc"] = roc_auc_score(y_test, y_proba)
metrics_before["average_precision"] = average_precision_score(y_test, y_proba)

metrics_after = classification_metrics(y_te_inv, y_pred_inv)
metrics_after["roc_auc"] = roc_auc_score(y_te_inv, y_proba_inv)
metrics_after["average_precision"] = average_precision_score(y_te_inv, y_proba_inv)

comparison_df = pd.DataFrame([metrics_before, metrics_after], index=["До додавання is_alone", "Після додавання is_alone"])
display(comparison_df.round(3))
```

**Висновок до індивідуального завдання:**  
Створено нову ознаку `is_alone`, яка показує, чи подорожував пасажир самостійно (`family_size == 1`). 
Після повторного навчання моделі та порівняння метрик якості видно, що додавання цієї ознаки дає змогу моделі краще враховувати фактор наявності супроводжуючих осіб при евакуації, що відображається у зміні точності (`precision`), повноти (`recall`) та загальної площі під ROC-кривою (`ROC AUC`).
